# 00 · Overview: how this project is built

This folder breaks the logistics-domain (M2) implementation of the CT30A1200 QAOA study into one
notebook per implementation step. Each notebook has the same structure:

- **What it does**: the implementation in one paragraph.
- **Why it was implemented**: the design reason, tied to TQ1 (formulation), TQ2 (implementation) or TQ3 (practical challenges).
- **Resources**: the repository code, libraries, data and literature it relies on.
- **Walk-through**: runnable cells that call the project code (nothing is re-implemented here).
- **Expected results**: the numbers you should see, followed by `assert` checks, so a notebook that runs to the end has verified its own claims.

| Notebook | Implementation | Code | TQ |

## Environment check
**What / why:** results are only reproducible with the same library versions. `requirements.txt` pins every
package; this cell compares the installed versions with the pins (the same check as `tests/test_env.py`).

```python
import platform
from importlib.metadata import version

pins = dict(line.split("==") for line in Path("requirements.txt").read_text().splitlines() if "==" in line)
mismatched = {name: (pin, version(name)) for name, pin in pins.items() if version(name) != pin}
print("Python", platform.python_version())
for name in ("qiskit", "qiskit-aer", "qiskit-optimization", "dimod", "dwave-samplers", "numpy", "scipy"):
    print(f"  {name:<20} {version(name)}")
print("packages differing from requirements.txt:", mismatched or "none")
```

**Expected results:** Python 3.12–3.14 and no mismatches. On Windows with Smart App Control enabled, an import
can occasionally fail with *"An Application Control policy has blocked this file"*; re-running the cell
usually succeeds (see `README.md`).
